```python
import numpy as np
```

```python
np.random.seed(2112)
X = np.random.randint(10, 101, size=(5, 5))
mu = np.mean(X)
s = np.std(X)

X_normalized = (X - mu) / s
Mean_normalized = np.mean(X_normalized).round()
Std_normalized = np.std(X_normalized).round()

print(F"value of X is:\n{X}\n")
print(F"value of X_normalized is:\n{X_normalized}\n")
print(F"Value of normalized mean is: {Mean_normalized}\n")
print(F"value of normalized standard deviation is: {Std_normalized}")

np.save("X_normalized.npy", X_normalized)
```

```python
C = (np.arange(1, 101) ** 3).reshape(10, 10)
div_by_4 = C[C % 4 == 0]

print(F"Shape of C: {C.shape}\n")
print(F"Array div_by_4:\n{div_by_4}\n")
print(F"The number of selected elements: {len(div_by_4)}")

np.save("div_by_4.npy", div_by_4)
```

```python
S = (np.arange(1, 37) ** 2).reshape(6,6)
S_mean = np.mean(S)
above_mean = S[S > S_mean]

print(F"Array S:\n{S}\n")
print(F"S_mean: {S_mean}\n")
print(F"Array above_mean:\n{above_mean}\n")
print(F"The number of selected elements: {len(above_mean)}")

np.save("above_mean.npy", above_mean)
```
